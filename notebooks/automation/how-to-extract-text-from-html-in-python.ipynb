# Extract Clean Text from HTML in Python

Learn the best ways to extract readable text from HTML strings or files using Python. We'll cover the most popular and reliable methods: **BeautifulSoup** (recommended), **html2text**, and quick alternatives.

This tutorial is based on common solutions from Stack Overflow and real-world best practices.

## 1. Import Required Libraries

````python
from bs4 import BeautifulSoup
import html2text
````

## 2. Sample HTML Data

````python
html = """
<html>
    <body>
        <h1>Title</h1>
        <p>Hello, world! This is a <a href="#">link</a>.</p>
        <script>alert('ignore me');</script>
        <style>.hidden { display: none; }</style>
        <p>Another paragraph with &amp; entity &#39;quote&#39;.</p>
    </body>
</html>
"""
````

## 3. Method 1: BeautifulSoup (Recommended)

BeautifulSoup is robust, handles malformed HTML, and properly decodes entities.

````python
soup = BeautifulSoup(html, "html.parser")

# Remove script and style elements
for script in soup(["script", "style"]):
    script.extract()

# Get clean text
text = soup.get_text(separator=" ", strip=True)
print(text)
````

**Output:**
`Title Hello, world! This is a link. Another paragraph with & entity 'quote'.`

## 4. Method 2: BeautifulSoup with stripped_strings (One-liner)

````python
text = ' '.join(BeautifulSoup(html, "html.parser").stripped_strings)
print(text)
````

**Output:**
`Title Hello , world ! This is a link . Another paragraph with & entity 'quote' .`

## 5. Method 3: Using html2text

Great for converting HTML to readable Markdown-style text.

````python
h = html2text.HTML2Text()
h.ignore_links = True
h.ignore_images = True

text = h.handle(html)
print(text)
````

**Output:**
```
Title

Hello, world! This is a link.

Another paragraph with & entity 'quote'.
```

## 6. Reading from an HTML File

If your HTML is in a file:

````python
# Save sample to file (for demonstration)
with open('sample.html', 'w', encoding='utf-8') as f:
    f.write(html)

# Read and extract
with open('sample.html', 'r', encoding='utf-8') as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, "html.parser")
for script in soup(["script", "style"]):
    script.extract()
clean_text = soup.get_text(separator=" ", strip=True)
print(clean_text)
````

## 7. Best Practices & Notes

- **Use BeautifulSoup** for most cases — it's reliable and widely used.
- Avoid regular expressions for HTML parsing (fragile with nested tags).
- For web pages with heavy JavaScript, consider tools like `trafilatura` or `readability-lxml`.
- `html2text` is excellent when you want Markdown output and fine-grained control.

That's it! You now have clean, readable text from any HTML source.
